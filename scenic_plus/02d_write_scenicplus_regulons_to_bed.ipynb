```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import subprocess
```

```python
projdir = '/u/project/cluo/terencew/igvf/2023_YR2/multiome'
```

```python
d_regulon = pd.read_csv(f'{projdir}/csv/gene_networks/scenic/direct_e_regulon.csv', sep='\t', index_col=0)
e_regulon = pd.read_csv(f'{projdir}/csv/gene_networks/scenic/extended_e_regulon.csv', sep='\t', index_col=0)
d_regulon.shape, e_regulon.shape
```

```python
d_regulon
```

```python
d_regulon.head()
```

```python
d_bed = pd.DataFrame(index=d_regulon['Region'])
d_bed['chrom'] = [x.split(':')[0] for x in d_bed.index]
d_bed['start'] = [int(x.split(':')[1].split('-')[0]) for x in d_bed.index]
d_bed['end'] = [int(x.split(':')[1].split('-')[1]) for x in d_bed.index]
d_bed.sort_values(by=['chrom', 'start'], inplace=True)

###
autosome = [f'chr{x}' for x in range(1, 23)]
mask = [x in autosome for x in d_bed['chrom']]
d_bed = d_bed[mask]

###
d_bed = d_bed[~d_bed.index.duplicated()]
d_bed['chr_num'] = [int(x.split('chr')[1]) for x in d_bed['chrom']]
d_bed.sort_values(by=['chr_num', 'start'], inplace=True)
d_bed.shape
```

```python
e_bed = pd.DataFrame(index=e_regulon['Region'])
e_bed['chrom'] = [x.split(':')[0] for x in e_bed.index]
e_bed['start'] = [int(x.split(':')[1].split('-')[0]) for x in e_bed.index]
e_bed['end'] = [int(x.split(':')[1].split('-')[1]) for x in e_bed.index]
e_bed.sort_values(by=['chrom', 'start'], inplace=True)

###
autosome = [f'chr{x}' for x in range(1, 23)]
mask = [x in autosome for x in e_bed['chrom']]
e_bed = e_bed[mask]

###
e_bed = e_bed[~e_bed.index.duplicated()]
e_bed['chr_num'] = [int(x.split('chr')[1]) for x in e_bed['chrom']]
e_bed.sort_values(by=['chr_num', 'start'], inplace=True)
e_bed.shape
```

```python
d_bed.head()
```

```python
e_bed.head()
```

```python
out = f'{projdir}/bed/scenic/dregulon.bed'
d_bed.to_csv(out, sep='\t', header=None, index=None)

subprocess.run(['bgzip', '-f', out], check=True)
subprocess.run(['tabix', '-f', '-p', 'bed', f'{out}.gz'], check=True)
```

```python
out = f'{projdir}/bed/scenic/eregulon.bed'
e_bed.to_csv(out, sep='\t', header=None, index=None)

subprocess.run(['bgzip', '-f', out], check=True)
subprocess.run(['tabix', '-f', '-p', 'bed', f'{out}.gz'], check=True)
```
