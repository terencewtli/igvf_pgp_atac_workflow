```r
suppressPackageStartupMessages(library(ArchR))
```

```r
addArchRGenome("hg38")
addArchRThreads(threads = 50)
```

```r
projdir <- '/u/project/cluo/terencew/igvf/2023_YR2/multiome/'
samples <- read.table(paste0(projdir, 'txt/samples.txt'), sep='\t')$V1
donors <- read.table(paste0(projdir, 'txt/donors.txt'), sep='\t')$V1
```

```r
frag_dir = paste0(projdir, 'atac/fragments/')
frag_paths <- c()
for (sample in samples) {
    frag_paths <- c(frag_paths, paste0(frag_dir, sample,'.fragments.tsv.gz'))
}
names(frag_paths) <- samples
inputFiles <- frag_paths
inputFiles
```

```r
outdir <- paste0(projdir, '/atac/arrow/ambimux/wgs/')
setwd(outdir)
getwd()
```

```r
ArrowFiles <- createArrowFiles(
  inputFiles = inputFiles,
  sampleNames = names(inputFiles),
  filterTSS = 0,
  filterFrags = 0,
  addTileMat = TRUE,
  addGeneScoreMat = TRUE,
)
```

```r
arrow_dir = paste0(projdir, '/atac/arrow/ambimux/wgs/')
ArrowFiles <- c()
for (sample in samples) {
    ArrowFiles <- c(ArrowFiles, paste0(arrow_dir, sample,'.arrow'))
}
names(ArrowFiles) <- samples
ArrowFiles
```

```r
outdir <- paste0(projdir, '/atac/archr/raw/YS3_ambimux')
proj <- ArchRProject(
  ArrowFiles = ArrowFiles, 
  outputDirectory = outdir,
  copyArrows = TRUE
)
```

```r
outdir
```

```r
meta <- as.data.frame(getCellColData(proj))
dim(meta)
```

```r
out <- paste0(projdir, '/csv/atac/archr/ambimux_atac_qc.csv')
write.table(meta, out, sep='\t', quote=FALSE)
```

```r
saveArchRProject(ArchRProj = proj,
                 outputDirectory = outdir,
                 load = FALSE)
```
